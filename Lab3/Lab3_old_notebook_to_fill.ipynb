## ROC: Reciever Operator Charectaristics 


```python
import matplotlib.pyplot as plt
from sklearn.metrics import roc_curve, roc_auc_score, precision_recall_curve, auc

# Find the optimal threshold for each model using Youden's J statistic
def get_optimal_threshold(fpr, tpr, thresholds):
    # Calculate Youden's J statistic for each threshold
    youden_j = tpr - fpr
    # Find the index of the maximum Youden's J statistic
    optimal_index = np.argmax(youden_j)
    # Return the optimal threshold and corresponding FPR and TPR
    return thresholds[optimal_index], fpr[optimal_index], tpr[optimal_index]
```



The Precision-Recall (PR) curve is a graphical representation of the trade-off between precision (positive predictive value) and recall (true positive rate) for different thresholds of a classification model. It is particularly useful when dealing with imbalanced datasets, where one class is significantly underrepresented compared to the other.

#### Key Points:
1. **Precision**: The proportion of true positive predictions out of all positive predictions made by the model.
    - Precision = True Positives / (True Positives + False Positives)

2. **Recall**: The proportion of true positive predictions out of all actual positive instances.
    - Recall = True Positives / (True Positives + False Negatives)

3. **AUC (Area Under the Curve)**: The area under the PR curve quantifies the overall performance of the model. A higher AUC indicates better performance, with a perfect model achieving an AUC of 1.0.

4. **Curve Shape**: 
    - A steep curve with high precision and recall indicates a good model.
    - A curve closer to the bottom-right corner indicates poor performance.

5. **Interpretation**:
    - The PR curve highlights the trade-off between precision and recall. A higher precision indicates fewer false positives, while a higher recall indicates fewer false negatives.
    - The curve helps in selecting an optimal threshold for the model, depending on whether precision or recall is more critical for the specific application.

In the plotted graph, the blue curve represents the PR curve for the baseline logistic regression model. The AUC value is displayed in the legend, which quantifies the model's ability to balance precision and recall effectively. The closer the curve is to the top-right corner, the better the model's performance.
